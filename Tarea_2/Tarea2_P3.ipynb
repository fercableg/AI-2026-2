### 3. (20 puntos, solo para grupos de a tres)

En el contexto de la inusual tormenta eléctrica registrada en Santiago, deberá abordar la identificación de estados meteorológicos y la detección de eventos climáticos extremos asumiendo una ausencia total de registros o confirmaciones visuales (*ground truth*). Para este ejercicio, el desafío consiste en descubrir la firma matemática de la tormenta basándose únicamente en la estructura de los datos. Seleccione un dataset meteorológico con al menos 15.000 observaciones temporales y 8 características de la zona (como presión, humedad, temperatura, etc.). Asegúrese de descartar cualquier variable objetivo o etiqueta, trabajando en un escenario puramente no supervisado.

Primero, dado que las variables atmosféricas suelen estar altamente correlacionadas, investigue y aplique una técnica de reducción de dimensionalidad espacial, justificando estadísticamente la cantidad de información retenida. Posteriormente, implemente dos enfoques distintos de agrupamiento (*clustering*) para descubrir los diferentes "estados" latentes del clima:

1. Una estrategia que busque particionar el espacio asignando las observaciones al centroide más cercano.
2. Una estrategia que agrupe los datos basándose en áreas de alta densidad, aislando las zonas dispersas.

Defina, aplique y justifique el uso de métricas de validación interna para evaluar la calidad geométrica de estas agrupaciones.

Finalmente, elabore una discusión técnica comparativa analizando:

* **a)** Considerando la distinta naturaleza matemática y magnitud de las variables atmosféricas (ej. presión en hPa frente a ráfagas de viento en km/h), discuta cómo el método utilizado para normalizar los datos (preprocesamiento) y la métrica de distancia seleccionados determinaron la forma final de los clústeres y la sensibilidad en la detección de anomalías.
* **b)** Aunque los algoritmos de agrupamiento evaluaron cada registro de forma independiente y sin noción del tiempo, analice si las observaciones identificadas como anomalías ocurrieron de forma contigua cronológicamente, evidenciando matemáticamente el ciclo de vida de la tormenta (formación, clímax y disipación).
